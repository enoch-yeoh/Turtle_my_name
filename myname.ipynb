```python
import turtle

# Screen setup
screen = turtle.Screen()
screen.setup(800, 400)
screen.bgcolor("black")
screen.title("Drawing ENOCH")

# Turtle setup
pen = turtle.Turtle()
pen.speed(5)
pen.color("white")
pen.pensize(3)

def draw_E(x, y, size):
    pen.penup()
    pen.goto(x, y)
    pen.pendown()
    pen.goto(x, y + size)  # Vertical line
    pen.goto(x + size/2, y + size)  # Top horizontal
    pen.penup()
    pen.goto(x, y + size/2)
    pen.pendown()
    pen.goto(x + size/2, y + size/2)  # Middle horizontal
    pen.penup()
    pen.goto(x, y)
    pen.pendown()
    pen.goto(x + size/2, y)  # Bottom horizontal

def draw_N(x, y, size):
    pen.penup()
    pen.goto(x, y)
    pen.pendown()
    pen.goto(x, y + size)  # Left vertical
    pen.goto(x + size/2, y)  # Diagonal
    pen.goto(x + size/2, y + size)  # Right vertical

def draw_O(x, y, size):
    pen.penup()
    pen.goto(x + size/4, y)
    pen.pendown()
    pen.circle(size/4)

def draw_C(x, y, size):
    pen.penup()
    pen.goto(x + size/2, y + size/4)
    pen.pendown()
    pen.setheading(180)
    pen.circle(size/4, 180)

def draw_H(x, y, size):
    pen.penup()
    pen.goto(x, y)
    pen.pendown()
    pen.goto(x, y + size)  # Left vertical
    pen.penup()
    pen.goto(x, y + size/2)
    pen.pendown()
    pen.goto(x + size/2, y + size/2)  # Horizontal
    pen.penup()
    pen.goto(x + size/2, y)
    pen.pendown()
    pen.goto(x + size/2, y + size)  # Right vertical

# Starting position
x = -200
y = -25
size = 50
spacing = 60

# Draw each letter
draw_E(x, y, size)
x += spacing
draw_N(x, y, size)
x += spacing
draw_O(x, y, size)
x += spacing
draw_C(x, y, size)
x += spacing
draw_H(x, y, size)

# Hide turtle and keep window open
pen.hideturtle()
screen.exitonclick()  # Click to close
```
